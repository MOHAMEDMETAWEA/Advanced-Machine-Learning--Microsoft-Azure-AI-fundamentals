```python
# Import required libraries
import tensorflow as tf
import numpy as np
import string

# Define input sequences
inputs = np.array([
    ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L", "M", "N", "O", "P", "Q", "R", "S"],
    ["Z", "Y", "X", "W", "V", "U", "T", "S", "R", "Q", "P", "O", "N", "M", "L", "K", "J", "I", "H"],
    ["B", "D", "F", "H", "J", "L", "N", "P", "R", "T", "V", "X", "Z", "A", "C", "E", "G", "I", "K"],
    ["M", "N", "O", "P", "Q", "R", "S", "T", "U", "V", "W", "X", "Y", "Z", "A", "B", "C", "D", "E"],
    ["H", "G", "F", "E", "D", "C", "B", "A", "L", "K", "J", "I", "P", "O", "N", "M", "U", "T", "S"],
])

# Define expected output sequences
expected = np.array([
    ["B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L", "M", "N", "O", "P", "Q", "R", "S", "T"],
    ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L", "M", "N", "O", "P", "Q", "R", "S"],
    ["C", "E", "G", "I", "K", "M", "O", "Q", "S", "U", "W", "Y", "A", "B", "D", "F", "H", "J", "L"],
    ["N", "O", "P", "Q", "R", "S", "T", "U", "V", "W", "X", "Y", "Z", "A", "B", "C", "D", "E", "F"],
    ["I", "J", "K", "L", "M", "N", "O", "P", "Q", "R", "S", "T", "U", "V", "W", "X", "Y", "Z", "A"],
])
```

```python
# Encode characters to numeric indices
input_encoded = np.vectorize(string.ascii_uppercase.index)(inputs)
input_encoded = input_encoded.astype(np.float32)
print(input_encoded)
```

```python
one_hot_inputs = tf.keras.utils.to_categorical(input_encoded)
print(one_hot_inputs)
```

```python
expected_encoded = np.vectorize(string.ascii_uppercase.index)(expected)
expected_encoded = expected_encoded.astype(np.float32)
one_hot_expected = tf.keras.utils.to_categorical(expected_encoded)

# Build the RNN model
model = tf.keras.Sequential([
    tf.keras.layers.SimpleRNN(128, return_sequences=True),
    tf.keras.layers.Dense(len(string.ascii_uppercase)),
])

# Compile the model
model.compile(loss="categorical_crossentropy", optimizer="adam")

# Train the model
model.fit(one_hot_inputs, one_hot_expected, epochs=100)
```

```python
# Inspect model weights
for i, layer in enumerate(model.layers):
    print(f"\nLayer {i + 1}: {layer.name}")
    weights = layer.get_weights()
    if weights:
        if "rnn" in layer.name.lower():
            print("RNN Layer Weights:")
            print(f"Input weights shape: {weights[0].shape}")
            print(f"Recurrent weights shape: {weights[1].shape}")
            print(f"Bias shape: {weights[2].shape}")
        elif "dense" in layer.name.lower():
            print("Dense Layer Weights:")
            print(f"Weights shape: {weights[0].shape}")
            print(f"Bias shape: {weights[1].shape}")

# Summarize the model
model.summary()
```

```python
# Make a prediction on new data
new_inputs = np.array([["B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L", "M", "N", "O", "P", "Q", "R", "S", "T"]])
new_inputs_encoded = np.vectorize(string.ascii_uppercase.index)(new_inputs)
new_inputs_encoded = new_inputs_encoded.astype(np.float32)
# Specify num_classes to ensure consistent one-hot encoding
new_inputs_encoded = tf.keras.utils.to_categorical(new_inputs_encoded, num_classes=len(string.ascii_uppercase))

# Predict the next character
prediction = model.predict(new_inputs_encoded)
predicted_index = np.argmax(prediction[0][-1])
print(f"Predicted next character: {string.ascii_uppercase[predicted_index]}")
```

```python
# Import required libraries
import tensorflow as tf
import numpy as np
import string

# Generate a larger dataset
alphabet = string.ascii_uppercase
dataset_size = 1000  # Increase dataset size
sequence_length = 20

inputs = []
expected = []

for _ in range(dataset_size):
    start = np.random.randint(0, len(alphabet) - sequence_length - 1)
    sequence = alphabet[start : start + sequence_length]
    next_char = alphabet[start + 1 : start + sequence_length + 1]
    inputs.append(list(sequence))
    expected.append(list(next_char))

inputs = np.array(inputs)
expected = np.array(expected)
```

```python
inputs[:2]
```

```python
# Encode characters to numeric indices and one-hot encode
def encode_sequences(data, alphabet):
    indices = np.vectorize(alphabet.index)(data)
    one_hot_encoded = tf.keras.utils.to_categorical(indices, num_classes=len(alphabet))
    return one_hot_encoded

one_hot_inputs = encode_sequences(inputs, alphabet)
one_hot_expected = encode_sequences(expected, alphabet)

# Build an improved RNN model
model = tf.keras.Sequential([
    tf.keras.layers.LSTM(128, return_sequences=True, input_shape=(sequence_length, len(alphabet))),
    tf.keras.layers.Dropout(0.2),  # Prevent overfitting
    tf.keras.layers.Dense(len(alphabet), activation="softmax"),
])

# Compile the model
model.compile(
    loss="categorical_crossentropy",
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    metrics=["accuracy"],
)

# Train the model
history = model.fit(
    one_hot_inputs,
    one_hot_expected,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    callbacks=[tf.keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True)],
)

# Inspect model summary
model.summary()

# Predict on new data
test_input = np.array([list("ABCDEFGHIJKLMNOPQRST")])  # Example input
test_one_hot = encode_sequences(test_input, alphabet)

prediction = model.predict(test_one_hot)
predicted_indices = np.argmax(prediction[0], axis=1)
predicted_chars = "".join([alphabet[idx] for idx in predicted_indices])

print(f"Input: {''.join(test_input[0])}")
print(f"Predicted Next Characters: {predicted_chars}")
```
